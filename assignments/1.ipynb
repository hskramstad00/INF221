````python
A = [8, 9, 2, 3, 7, 1]

def InsertionSort(A):
    print(" j  |  i  | key  | A")
    print("--------------------------------")
    for j in range(1, len(A)):
        key = A[j]
        i = j - 1
        print(f" {j+1}  |  {i+1}  |  {key}   | {A}")
        while i >= 0 and A[i] > key:
            A[i + 1] = A[i]
            i -= 1
        A[i + 1] = key
    print(f" Terminated      | {A}")
    print("--------------------------------")
    return A

b = InsertionSort(A)
````
