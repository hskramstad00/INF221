```python
# Insertion sort algorithm from lecture
# Added print statments for each repetion of the loop
# Also used 1-based indexing to match pseudocode
def insertion_sort(A):
    print('| j | | i | |key| | A  |')
    for j in range(1, len(A)):
        key = A[j]
        i = j - 1
        print(f'| {j+1} | | {i+1} | | {key} | | {A} |')
        while i >= 0 and A[i] > key:
            A[i + 1] = A[i]
            i = i - 1
        A[i + 1] = key
    print(f'|Terminated| | | | | |{A}|')
        
A = [8,9,2,3,7,1]
insertion_sort(A)
```
