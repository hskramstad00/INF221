## Exercise 1

General solution of the form, using the Master theorem:

$$
T(n) = aT\left(\frac{n}{b}\right) + c n^k
$$

This has three cases:

$$
T(n) =
\begin{cases}
\Theta\left(n^k\right) & \text{if } a < b^k \\[6pt]
\Theta\left(n^k \log n\right) & \text{if } a = b^k \\[6pt]
\Theta\left(n^{\log_b a}\right) & \text{if } a > b^k
\end{cases}
$$

#### a)

$$
\begin{aligned}
T(n) &= 2T\left(\frac{n}{4}\right) + 1 \\
a &= 2,\quad b = 4,\quad k = 0 \\
a &> b^k \\
T(n) &= \Theta\left(\sqrt{n}\right)
\end{aligned}
$$

#### b)

$$
\begin{aligned}
T(n) &= 2T\left(\frac{n}{4}\right) + \sqrt{n} \\
a &= 2,\quad b = 4,\quad k = \frac{1}{2} \\
a &= b^k \\
T(n) &= \Theta\left(\sqrt{n}\log n\right)
\end{aligned}
$$

#### c)

$$
\begin{aligned}
T(n) &= 2T\left(\frac{n}{4}\right) + n \\
a &= 2,\quad b = 4,\quad k = 1 \\
a &< b^k \\
T(n) &= \Theta(n)
\end{aligned}
$$

#### d)

$$
\begin{aligned}
T(n) &= 2T\left(\frac{n}{4}\right) + n^2 \\
a &= 2,\quad b = 4,\quad k = 2 \\
a &< b^k \\
T(n) &= \Theta\left(n^2\right)
\end{aligned}
$$

`````python
def heapify(arr, i, n):
    smallest = i
    l = 2 * i + 1
    r = 2 * i + 2

    # If left child exists and is smaller than root
    if l < n and arr[l] < arr[smallest]:
        smallest = l

    # If right child exists and is smaller than smallest so far
    if r < n and arr[r] < arr[smallest]:
        smallest = r

    # If smallest is not root,
    # swap and continue heapifying
    if smallest != i:
        arr[i], arr[smallest] = arr[smallest], arr[i]
        # Recursively heapify
        heapify(arr, smallest, n)

def sort(list_of_lists):
    # (value, list_idx, element_idx)
    arr = []
    for i, lst in enumerate(list_of_lists):
        if lst:
            arr.append((lst[0], i, 0))

    #  Build min-heap
    for i in range(len(arr) // 2 - 1, -1, -1):
        heapify(arr, i, len(arr))
    result = []
    while arr:
        val, list_idx, elem_idx = arr[0]
        result.append(val)

        next_idx = elem_idx + 1
        if next_idx < len(list_of_lists[list_idx]):
            arr[0] = (list_of_lists[list_idx][next_idx], list_idx, next_idx)
        else:
            arr[0] = arr[-1]
            arr.pop()
        # heapify 
        heapify(arr, 0, len(arr))

    return result


a = [1, 5, 7]
b = [2, 4]
c = [3, 6, 8, 9]

list_of_lists = [a, b, c]
sorted_list = sort(list_of_lists)
print(sorted_list)
`````
