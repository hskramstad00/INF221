## Exercise 1

General solution of the form, using the Master theorem:

$$
T(n) = aT\left(\frac{n}{b}\right) + c n^k
$$

This has three cases:

$$
T(n) =
\begin{cases}
\Theta\left(n^k\right) & \text{if } a < b^k \\[6pt]
\Theta\left(n^k \log n\right) & \text{if } a = b^k \\[6pt]
\Theta\left(n^{\log_b a}\right) & \text{if } a > b^k
\end{cases}
$$

#### a)

$$
\begin{aligned}
T(n) &= 2T\left(\frac{n}{4}\right) + 1 \\
a &= 2,\quad b = 4,\quad k = 0 \\
a &> b^k \\
T(n) &= \Theta\left(\sqrt{n}\right)
\end{aligned}
$$

#### b)

$$
\begin{aligned}
T(n) &= 2T\left(\frac{n}{4}\right) + \sqrt{n} \\
a &= 2,\quad b = 4,\quad k = \frac{1}{2} \\
a &= b^k \\
T(n) &= \Theta\left(\sqrt{n}\log n\right)
\end{aligned}
$$

#### c)

$$
\begin{aligned}
T(n) &= 2T\left(\frac{n}{4}\right) + n \\
a &= 2,\quad b = 4,\quad k = 1 \\
a &< b^k \\
T(n) &= \Theta(n)
\end{aligned}
$$

#### d)

$$
\begin{aligned}
T(n) &= 2T\left(\frac{n}{4}\right) + n^2 \\
a &= 2,\quad b = 4,\quad k = 2 \\
a &< b^k \\
T(n) &= \Theta\left(n^2\right)
\end{aligned}
$$

## Exercise 2

````
Permute-Without-Identity(A)
  n = A.length
  for i = 1 to n-1
    swap A[i] with A[Random(i+1, n)]
````

The algroithm  will never return the original ordring. That is because in the first step i=1, and it's swapped with some A[j], where j is from 2 to n element. Since all elements are unique, then A[1] will always be a different value then the original.

The algoritm will not generate all possibel permutations for a given array. At iteration i, the algorithm will choses form n-1 positions. The the total number of outcomes is:

(n-1) × (n-2) × ... × 1 = (n-1)!


Exampel:
**i = 1:** swap `A[1]` with `A[Random(2, 3)]` will be one of these choices:
- `j = 2` → `[2, 1, 3]`
- `j = 3` → `[3, 2, 1]`

**i = 2:** swap `A[2]` with `A[Random(3, 3)]` thet it's only once choice:
- `[2, 1, 3]` → `[2, 3, 1]`
- `[3, 2, 1]` → `[3, 1, 2]`

The algorithm produces 2 outcomes, each with probability 1/2
